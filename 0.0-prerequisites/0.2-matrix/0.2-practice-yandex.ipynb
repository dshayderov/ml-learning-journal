### Задание 1. Проверка типа матрицы

На вход программы подается квадратная матрица. Необходимо определить ее тип: диагональная, верхнетреугольная или нижнетреугольная. Если матрица не относится ни к одному из указанных типов, вывести `OTHER`. В противном случае вывести одну из строк: `DIAGONAL`, `UPPER_TRIANGULAR` или `LOWER_TRIANGULAR`.

```python
def main():
    n = int(input())
    if not (2 <= n <= 10):
        raise Exception("Введен неверный размер матрицы")

    matrix = np.array([
        [int(x) for x in input().split(" ")] for t in range(n)
    ])

    # индексы матрицы
    i, j = np.indices(matrix.shape)

    #Проверка того, что элементы выше главной диагонали равны 0
    upper_check = np.all(matrix[i < j] == 0)

    #Проверка того, что элементы ниже главной диагонали равны 0
    lower_check = np.all(matrix[i > j] == 0)

    status = ""
    if upper_check and lower_check:
        status = "DIAGONAL"
    elif upper_check and not lower_check:
        status = "LOWER_TRIANGULAR"
    elif lower_check and not upper_check:
        status = "UPPER_TRIANGULAR"
    else:
        status = "OTHER"

    print(status)


if __name__ == '__main__':
    main()
```

### Задание 2. Умножение матриц вручную

На вход программы подаются две вещественные матрицы $A$ и $B$. Необходимо самостоятельно реализовать их умножение и получить $C=AB$, если число столбцов $A$ совпадает с числом строк $B$. В противном случае вывести `NOT_DEFINED`. Использование встроенных средств `numpy` для умножения матриц не допускается.

```python
def main():
    # Ввод матрицы A
    [m, n] =[int(par) for par in input().split(" ")]
    if not (2 <= m <= 10 and 2 <= n <= 10):
        raise Exception("Введены неверные параметры m и n")
        
    A = np.array([[int(x) for x in input().split(" ")] for coor in range(m)])

    if not A.shape == (m, n):
       raise Exception("Размер матрицы не соответствует заданному")

    # Ввод матрицы B
    [h, k] =[int(par) for par in input().split(" ")]
    if not (2 <= h <= 10 and 2 <= k <= 10):
        raise Exception("Введены неверные параметры h и k")
        
    B = np.array([[int(x) for x in input().split(" ")] for coor in range(h)])

    if not B.shape == (h, k):
       raise Exception("Размер матрицы не соответствует заданному")

    # Проверка на согласованность
    if n != h:
        print("NOT_DEFINED")
    else:
        C = np.array([
            [np.dot(A[i, :], B[:, j]) for j in range(k)] 
            for i in range(m)
        ])

        print('\n'.join(' '.join(map(str, row)) for row in C))


if __name__ == '__main__':
    main()
```

### Задание 3. Нильпотентные матрицы

На вход программы подается квадратная матрица $A$. Необходимо найти наименьшее натуральное число $k$, при котором $A^k$ является нулевой матрицей, и вывести это значение $k$.

```python
def main():
    # Ввод матрицы
    n = int(input())   
    if not (1 <= n <= 10):
        raise Exception("Введен неверный размер матрицы")

    matrix = np.array([
        [int(x) for x in input().split(" ")] for t in range(n)
    ])

    # Нахождение k
    k = 1
    for i in range(1, 101):
        pow_matrix = np.linalg.matrix_power(matrix, i)
        if np.all(pow_matrix == 0):
            k = i
            break

    print(k)    

if __name__ == '__main__':
    main()
```

### Задание 5. Нормализация столбцов матрицы

На вход программы подается вещественная матрица размера $m\times n$. Для каждого значения столбца необходимо вычесть его среднее значение, разделить результат на стандартное отклонение и отбросить дробную часть полученных значений. Вывести нормализованную матрицу.

```python
def main():
    # Ввод матрицы A
    [m, n] =[int(par) for par in input().split(" ")]
    if not (2 <= m <= 10 and 2 <= n <= 10):
        raise Exception("Введены неверные параметры m и n")
        
    A = np.array([[int(x) for x in input().split(" ")] for coor in range(m)])

    if not A.shape == (m, n):
       raise Exception("Размер матрицы не соответствует заданному")

    # Нормализация матрицы
    A_norm = A.copy()

    for i in range(n):
        col_mean = np.mean(A_norm[:,i])
        col_std = np.std(A_norm[:,i])
        A_norm[:,i] = np.array((A_norm[:,i] - col_mean) / col_std).astype(int)
        
    print('\n'.join(' '.join(map(str, row)) for row in A_norm))


if __name__ == '__main__':
    main()
```

### Задание 1. Обратная матрица через присоединённую матрицу

На вход программы подается обратимая целочисленная матрица $A$ размера $n\times n$. Необходимо найти её обратную матрицу, используя присоединённую матрицу и определитель:

$$A^{-1}=\frac{\operatorname{adj}(A)}{\det A}.$$

Вывести полученную обратную матрицу в виде списка из $n$ строк по $n$ элементов.

```python
def main():
    # Ввод матрицы A
    n = int(input())   
    if not (1 <= n <= 10):
        raise Exception("Введен неверный размер матрицы")

    A = np.array([
        [int(x) for x in input().split()] for t in range(n)
    ])

    # Присоединенная матрица
    B = np.empty((n, n))
    for i in range(n):
        for j in range(n):
            # Вычисляем алгебраические дополнения
            rows = [row for row in range(A.shape[0]) if row != i]
            cols = [col for col in range(A.shape[1]) if col != j]
            b_ij = (-1) ** (i + j) * np.linalg.det(A[np.ix_(rows, cols)])

            # Заполняем присоединенную матрицу
            B[j, i] = b_ij

    # Вычисляем определитель матрицы A
    det_A = np.linalg.det(A)
    
    # Находим обратную матрицу
    A_inv = np.round(np.array(B / det_A)).astype(int)

    print('\n'.join(' '.join(map(str, row)) for row in A_inv))

if __name__ == '__main__':
    main()
```

### Задание 2. Определитель специальной матрицы

На вход программы подаются число $n$ и $n$ целых чисел $x_1,\ldots,x_n$, задающих матрицу Вандермонда:

$$A_{ij}=x_i^{\,j-1}.$$

Необходимо вычислить $\det A$ за $O(n^2)$, используя структуру матрицы, и вывести результат, округлённый до целого числа.

```python
def main():
    # Ввод чисел
    n = int(input())   
    if not (1 <= n <= 10):
        raise Exception("Введен неверный размер матрицы")

    x = np.array([int(x) for x in input().split()])

    # Вычисление определителя
    det = 1
    for i in range(n):
        for j in range(i + 1, n):
            det *= x[j] - x[i]

    print(round(det))

if __name__ == '__main__':
    main()
```

### Задание 3. Детерминант с параметром

На вход подаются семь целых чисел, задающих параметры матрицы $3\times3$, элементы которой линейно зависят от параметра $x$. Необходимо найти все целые значения $x$, при которых $\det A=0\$. Корни с кратностью больше одной нужно вывести соответствующее число раз, округлить до целого и отсортировать по возрастанию. Если подходящих корней нет, вывести `NO_ROOTS`.

```python
import math
```

```python
def main():
    # Ввод чисел
    [a,p,q,r,s,t,u] = [int(x) for x in input().split()]
    if not all(-100 <= _ <= 100 for _ in [a,p,q,r,s,t,u]):
        raise Exception("Параметры не соответствуют условию")

    # Приводим разложение определителя к квадратному уравнению
    # det A = ax^2 - (pr + qt)x + (pst + qru - asu)
    A = a
    B = -(p*r + q*t)
    C = p*s*t + q*r*u - a*s*u

    # Список корней
    roots = []

    if A == 0:
        # Линейное уравнение Bx + C = 0
        if B != 0:
            if (-C) % B == 0:
                roots.append((-C) // B)

    else:
        # Квадратное уравнение Ax^2 + Bx + C = 0
        D = B ** 2 - 4 * A * C

        if D >= 0:
            sqrt_D = math.isqrt(D)

            # Поиск корней
            if sqrt_D ** 2 == D:
                num1 = -B + sqrt_D
                num2 = -B - sqrt_D
                den = 2 * A

                if num1 % den == 0:
                    roots.append(num1 // den)

                if num2 % den == 0:
                    roots.append(num2 // den)

    if roots:
        print(*sorted(roots))
    else:
        print("NO_ROOTS")
        

if __name__ == '__main__':
    main()
```

### Задание 1. Сжатие изображения через усечённое SVD

На вход подаётся матрица яркости $A\in\mathbb{R}^{n\times m}$ со значениями от 0 до 1 и несколько значений $k$. Для каждого $k$ необходимо построить SVD-приближение матрицы ранга $k$, оставив первые $k$ сингулярных значений, и вычислить относительную ошибку по норме Фробениуса и степень сжатия:

$$
\frac{\|A-A_k\|_F}{\|A\|_F},
\qquad
\frac{k(n+m+1)}{nm}.
$$

Для каждого $k$ вывести соответствующую пару значений с точностью не менее четырёх знаков после запятой.

```python
def main():
    # Ввод матрицы A
    [n, m] =[int(par) for par in input().split()]
    if not (2 <= m <= 10 and 2 <= n <= 10):
        raise Exception("Введены неверные параметры m и n")
        
    A = np.array([[float(x) for x in input().split()] for coor in range(n)])

    if not A.shape == (n, m):
       raise Exception("Размер матрицы не соответствует заданному")

    # Ввод параметров t и списка значений k
    t = int(input())
    k_list = [int(x) for x in input().split()]

     # Вычисляем SVD
    U, s, Vt = np.linalg.svd(A, full_matrices=False)

    for k in k_list:
        k_eff = min(k, n, m)
        
        # Строим низкоранговое приближение
        A_k = (U[:, :k_eff] * s[:k_eff]) @ Vt[:k_eff, :]

        # Относительная ошибка по норме Фробениуса
        fro_error = np.linalg.norm(A - A_k, ord='fro') / np.linalg.norm(
            A, ord='fro')
        
        # Степень сжатия
        compress_rate = (k_eff*(n+m+1)) / (n*m)

        print(fro_error, compress_rate)


if __name__ == '__main__':
    main()
```

### Задание 2. Решение системы $Ax=b$ методом Холецкого

На вход подаются симметричная матрица $A$ и вектор $b$. Необходимо решить систему $Ax=b$ с помощью разложения Холецкого $A=LL^T$, последовательно решив системы $Ly=b$ и $L^Tx=y$. Затем вычислить невязку $\|Ax-b\|_2$. Если разложение Холецкого невозможно, вывести `FAIL` и `nan`. Решение и невязку вывести с требуемой точностью.

```python
def main():
    # Ввод матрицы A
    n = int(input())   
    if not (1 <= n <= 10):
        raise Exception("Введен неверный размер матрицы")

    A = np.array([
        [int(x) for x in input().split()] for t in range(n)
    ])

    # Ввод вектора b
    b = [int(x) for x in input().split()]
    if not len(b) == n:
        raise Exception("Размер вектора b не соответствует заданному")

    try:
        # Разложение Холецкого: A = L * L^T
        L = np.linalg.cholesky(A)
        
        # Решаем L * y = b
        y = np.linalg.solve(L, b)
        
        # Решаем L^T * x = y
        x = np.linalg.solve(L.T, y)

        # Вычисляем невязку
        residual = np.linalg.norm(A @ x - b)
        
        print(" ".join(f"{num:.4f}" for num in x))
        print(f"{residual:.6f}")
    
    except np.linalg.LinAlgError:
        print("FAIL")
        print("nan")


if __name__ == '__main__':
    main()
```

### Задание 3. Собственные значения и спектральный радиус

На вход подается квадратная вещественная матрица $A$. Необходимо найти все её собственные значения и спектральный радиус $\rho(A)$ — максимальный модуль среди собственных значений. Собственные значения можно вывести в любом порядке с точностью до 4 знаков после запятой, а спектральный радиус — с точностью до 6 знаков.

```python
def main():
    # Ввод матрицы A
    n = int(input())   
    if not (1 <= n <= 10):
        raise Exception("Введен неверный размер матрицы")

    A = np.array([
        [int(x) for x in input().split()] for t in range(n)
    ])

    # Находим собственные значения
    eigenvalues = np.linalg.eigvals(A)

    # Приводим к типу float, отбрасывая мнимую часть
    float_eigenvalues = eigenvalues.real.astype(float)
    
    # Вычисляем спектральный радиус
    spectral_radius = np.max(np.abs(eigenvalues))

    print(" ".join(f"{eigen:.4f}" for eigen in float_eigenvalues))
    print(f"{spectral_radius:.6f}")

if __name__ == '__main__':
    main()
```
