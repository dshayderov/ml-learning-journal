### Задание 1. Вероятность выпадения не менее $K$ орлов

Нечестную монету с вероятностью орла $p$ подбрасывают $N$ раз. Необходимо вычислить вероятность того, что орёл выпадет не менее $K$ раз. Для построения пространства элементарных исходов одного подбрасывания и последовательностей из $N$ подбрасываний использовать `itertools.product`. Значение вероятности вывести с абсолютной погрешностью не более $10^{-5}$.

```python
import itertools
```

```python
def main():
    # Считывание данных
    [n, k] = [int(_) for _ in input().split(" ")]
    p = float(input())

    # Вероятность выпадения решки
    q = 1 - p

    # Список всех возможных элементарных исходов
    omega = itertools.product(["o", "r"], repeat=n)

    probability = sum(
        # Расчет суммарной верояности всех подходящих исходов
        (p ** item.count("o")) * (q ** item.count("r"))
        for item in omega
        # Отбор исходов, где орел выпадает не менее K раз
        if item.count("o") >= k
    )

    print(probability)


if __name__ == '__main__':
    main()
```

### Задание 2. Геометрическая вероятность

На вход подаются вещественные числа $C$ и $D$. Случайно и независимо выбираются $X$ и $Y$ из отрезка $[0,C]$. Необходимо вычислить вероятность выполнения условия $X+Y<D$ и вывести результат с точностью не менее $10^{-5}$.

```python
def main():
    # Считывание данных
    [C, D] = [float(_) for _ in input().split(" ")]

    if D <= 0:
        # X и Y больше 0, сумма не может быть отрицательной
        probability = 0.0
    elif D <= C:
        # Благоприятная область - прямоугольный треугольник с катетами D
        probability = D**2 / (2 * C**2)
    elif D < 2 * C:
        # Неблагоприятная область - треугольник с катетами 2C−D
        probability = 1 - (2 * C - D)**2 / (2 * C**2)
    else:
        # D > 2C, больше чем максимально возможная сумма X и Y
        probability = 1.0

    print(probability)


if __name__ == '__main__':
    main()
```

### Задание 3. Вероятности при извлечении шаров

В урне находятся $R$ красных, $G$ зелёных и $B$ синих шаров. Из неё без возвращения выбирают 3 шара. Необходимо вычислить вероятность того, что среди выбранных шаров есть хотя бы один зелёный, а также вероятность того, что все три шара имеют один цвет. Вывести обе вероятности через пробел с точностью не менее 5 знаков после запятой.

```python
def combinations(n):
    """Функция сочетания:
    C_kn = n! / k!(n-k)!
    k = 3 (R, G, B)"""
    if n < 3:
        return 0
    return n * (n - 1) * (n - 2) // 6
    
def main():
    # Считывание данных
    [R, G, B] = [int(_) for _ in input().split(" ")]

    # Все возможные комбинации
    N = combinations(R+G+B)

    # Вероятность, что хотя бы 1 шар будет зеленым
    # 1 - вероятность, что шар будет синим или красным
    P_green = 1 - combinations(R+B) / N

    # Вероятность, что все шары будут разными
    # Сумма трех взаимоисключающих событий
    P_diff = (combinations(R) + combinations(G) + combinations(B)) / N

    print(f"{P_green} {P_diff}")


if __name__ == '__main__':
    main()
```

### Задание 4. Проверка честности монетки

На вход подаются число подбрасываний $N$, требуемый уровень доверия $I$ и последовательность из $N$ результатов, где `1` означает орла, а `0` — решку. Необходимо найти наибольшее $K<N/2$, для которого вероятность получить число орлов в интервале $[K,N-K]$ при честной монете не меньше $I$. Затем определить, попадает ли фактическое число орлов в этот интервал: если да, вывести `fair`, иначе — `biased`. В первой строке вывести $K$ и $N-K$, во второй — результат проверки.

```python
import math
```

```python
def main():
    # Считывание первой строки
    first_input = input().split(" ")
    N = int(first_input[0])
    I = float(first_input[1])

    # Считывание второй строки
    exp = input().split(" ")

    K_result = 0
    # Поиск K
    for K in range(N // 2 + 1):
        probability = 0
    
        for k in range(K, N - K + 1):
            probability += math.comb(N, k) / 2**N
    
        if probability >= I:
            K_result = K

    print(f"{K_result} {N - K_result}")

    # Подсчет орлов в серии подбрасываний
    counter = exp.count("1")
    # Определение честности монетки
    if K_result <= counter <= N-K_result:
        print("fair")
    else:
        print("biased")


if __name__ == '__main__':
    main()
```

### Задание 1. Вероятность заболевания по результатам двух тестов

На вход подаются априорная вероятность заболевания $p$, чувствительность и вероятность ложноположительного результата для двух тестов $s_1,f_1,s_2,f_2$, а также результаты обоих тестов $test_1,test_2$. Предполагается, что результаты тестов условно независимы при фиксированном состоянии пациента. Необходимо по формуле Байеса вычислить вероятность заболевания при наблюдаемых результатах тестирования и вывести её с точностью не менее 5 знаков после запятой.

```python
def main():
    # Считывание данных
    [p, s1, f1, s2, f2] = [float(_) for _ in input().split(" ")]
    [test1, test2] = [int(_) for _ in input().split(" ")]

    # Тест 1
    if test1 == 1:
        # Вероятность положительного результата, когда человек реально болен
        p_test1_sick = s1
        # Вероятность положительного результата, когда человек здоров
        p_test1_healthy = f1
    else:
        # Вероятность отрицательного результата, когда человек болен
        p_test1_sick = 1 - s1
        # Вероятность отрицательного результата, когда человек реально здоров
        p_test1_healthy = 1 - f1

    # Тест 2
    if test2 == 1:
        # Вероятность положительного результата, когда человек реально болен
        p_test2_sick = s2
        # Вероятность положительного результата, когда человек здоров
        p_test2_healthy = f2
    else:
        # Вероятность отрицательного результата, когда человек болен
        p_test2_sick = 1 - s2
        # Вероятность отрицательного результата, когда человек реально здоров
        p_test2_healthy = 1 - f2

    # Вероятность быть больным при наблюдаемых результатах теста 1 и 2
    p_tests_sick = p_test1_sick * p_test2_sick
    # Вероятность быть здоровым при наблюдаемых результатах теста 1 и 2
    p_tests_healthy = p_test1_healthy * p_test2_healthy

    # P(Болен|test1, test2) = P(test1, test2|Болен)*P(Болен) /
    # / P(test1, test2|Болен)*P(Болен) + P(test1, test2|Не болен)*P(Не болен)
    result = (p_tests_sick * p) / (p_tests_sick * p + p_tests_healthy * (1 - p))

    print(result)


if __name__ == '__main__':
    main()
```

### Задание 2. Проверка независимости событий

На вход подаются результаты $M$ экспериментов для трёх событий $A$, $B$ и $C$, представленные нулями и единицами. По этим данным необходимо оценить вероятности $P(A)$, $P(B)$, $P(C)$, вероятности попарных пересечений и $P(A\cap B\cap C)$. Затем проверить попарную независимость событий и их независимость в совокупности. Вывести `ALL_INDEPENDENT`, если выполняются оба условия, `PAIRWISE_ONLY`, если пары независимы, но тройка нет, и `NOT_INDEPENDENT`, если хотя бы одна пара зависима.

```python
def main():
    # Считывание данных
    M = int(input())
    exp = [[int(_) for _ in input().split(" ")] 
           for i in range(M)]

    # Выделяем каждое из событий в отдельную группу
    A = [event[0] for event in exp]
    B = [event[1] for event in exp]
    C = [event[2] for event in exp]

    # Вычисляем вероятность каждого из событий
    # P = кол-во положительных исходов / кол-во всех исходов
    prob_A = A.count(1) / M
    prob_B = B.count(1) / M
    prob_C = C.count(1) / M

    # Определяем независимость событий
    # P(X∩Y)==P(X)P(Y)
    indep_AB = True if (sum(1 for a, b in zip(A, B) 
                            if a == b == 1)) / M == prob_A * prob_B else False
    indep_AC = True if (sum(1 for a, c in zip(A, C) 
                            if a == c == 1)) / M == prob_A * prob_C else False
    indep_BC = True if (sum(1 for b, c in zip(B, C) 
                            if b == c == 1)) / M == prob_B * prob_C else False
    indep_ABC = True if (sum(1 for a, b, c in zip(A, B, C) 
                             if a == b == c == 1)) / M == prob_A * prob_B * prob_C else False

    # Выводим результат независимости
    if all([indep_AB, indep_AC, indep_BC, indep_ABC]):
        print("ALL_INDEPENDENT")
    elif all([indep_AB, indep_AC, indep_BC]):
        print("PAIRWISE_ONLY")
    else:
        print("NOT_INDEPENDENT")


if __name__ == '__main__':
    main()
```

### Задание 3. Наивный байесовский классификатор

На вход подаются $M$ размеченных тренировочных писем, где первый признак указывает на спам, а остальные $K$ признаков показывают наличие ключевых слов, а также $N$ тестовых писем с теми же $K$ признаками. Необходимо построить наивный байесовский классификатор и для каждого тестового письма определить класс с большей условной вероятностью. Вывести `1` для спама, `0` для обычного письма и `-1`, если вероятности обоих классов совпадают или классификатор не позволяет однозначно определить класс. При сравнении использовать числители вероятностей без общего знаменателя.

```python
import math
```

```python
def main():
    # Считывание данных
    [M, N, K] = [int(_) for _ in input().split(" ")]
    train_letters = [[int(_) for _ in input().split(" ")] 
                     for i in range(M)]
    test_letters = [[int(_) for _ in input().split(" ")] 
                     for i in range(N)]

    # Разбиваем тренировочный набор на спам / не спам
    train_spam = [letter for letter in train_letters if letter[0] == 1]
    train_not_spam = [letter for letter in train_letters if letter[0] == 0]

    # Рассчитываем вероятность того, что письмо будет или не будет спамом
    P_spam = len(train_spam) / M
    P_not_spam = len(train_not_spam) / M

    # Вероятность того, что письмо будет или не будет спамом, если в нем есть ключ. слово
    # Собираем массив из вероятностей для каждого ключевого слова
    P_spam_keyword = [[letter[i] for letter in train_spam
                      ].count(1) / len(train_spam) for i in range(1, K + 1)
                     ]
    P_not_spam_keyword = [[letter[i] for letter in train_not_spam
                          ].count(1) / len(train_not_spam) for i in range(1, K + 1)
                         ]

    # Перебираем тестовый набор
    for letter in test_letters:
        P_letter_spam = 1
        P_letter_not_spam = 1
        for i, value in enumerate(letter):
            if value == 1:
                P_letter_spam *= P_spam_keyword[i]
                P_letter_not_spam *= P_not_spam_keyword[i]
            else:
                P_letter_spam *= 1 - P_spam_keyword[i]
                P_letter_not_spam *= 1 - P_not_spam_keyword[i]

        P_letter_spam *= P_spam
        P_letter_not_spam *=  P_not_spam

        if P_letter_spam > P_letter_not_spam:
            print("1", end=" ")
        elif P_letter_not_spam > P_letter_spam:
            print("0", end=" ")
        else:
            print("-1", end=" ")


if __name__ == '__main__':
    main()
```

### Задание 1. Распределение времени до второго успеха

На вход подаются число неудачных испытаний $N$ перед вторым успехом и вероятность успеха $p$. Необходимо вычислить вероятность того, что второй успех произойдет на $(N+1)$-м испытании в последовательности независимых испытаний. Результат вывести с абсолютной погрешностью не более $10^{-5}$.

```python
def main():
    # Считывание данных
    input_s = input().split(" ")
    N = int(input_s[0])
    p = float(input_s[1])

    # Рассчитываем вероятность второго успеха на N+1 испытании
    # Вероятность успеха на N+1 испытании = p 
    # p(1-p)^{N-1} - вероятность того, что за N испытаний было до одно успешное
    # Np(1-p)^{N-1} - так как любое из N испытаний могло быть успешным
    # Итоговая вероятность: pNp(1-p)^{N-1} = Np^2(1-p)^{N-1}
    probability = N * pow(p, 2) * pow(1 - p, N - 1)

    print(probability)


if __name__ == '__main__':
    main()
```

Используя соответствующее распределение из `scipy.stats`, необходимо вычислить вероятность того, что случайная величина принимает значение $N$, то есть значение функции вероятности `pmf(N)`. Результат вывести с абсолютной погрешностью не более $10^{-5}$.

```python
from scipy.stats import binom
from scipy.stats import poisson
from scipy.stats import geom
```

```python
def main():
    # Считывание данных
    stat = input()
    params = input()
    N = int(input())

    # Вероятность, что величина примет значение N
    probability = 0

    if stat == "биномиальное":
        # Биномальное распределение
        [n, p] = [float(_) for _ in params.split(" ")]
        probability = binom.pmf(N, n, p)
    elif stat == "геометрическое":
        # Геометрическое распределение
        p = float(params)
        probability = geom.pmf(N, p)
    else:
        # Распределение Пуассона
        lam = float(params)
        probability = poisson.pmf(N, lam)

    print(probability)


if __name__ == '__main__':
    main()
```

### Задание 3. Распределение суммы трёх кубиков

На вход подаётся целое число $N$. Необходимо вычислить вероятность того, что сумма очков при броске трёх независимых честных шестигранных кубиков равна $N$. Результат вывести с абсолютной погрешностью не более $10^{-5}$.

```python
def main():
    # Считывание данных
    N = int(input())

    probability = 0
    # Сумма очков для трех кубиков может варьироваться в интервале [3, 18],
    # для иных сумм вероятность выпадения равна 0
    if 3 <= N <= 18:
        # Счетчик кол-ва комбинаций с нужной суммой
        count = 0

        # Перебираем значения первого кубика
        for a in range(1, 7):
            # Перебираем значения второго кубика
            for b in range(1, 7):
                # На третьем кубике должен быть остаток для набора суммы
                c = N - a - b

                # Остаток должен вписываться в количество очков на кубике
                if 1 <= c <= 6:
                    count += 1

        # Итоговая вероятность: кол-во подходящих комбинаций / кол-во всех комбинаций
        probability = count / 6**3

    print(probability)


if __name__ == '__main__':
    main()
```

### Задание 2. Функция распределения смеси

На вход подаются параметры $N$ нормальных распределений: их веса $\alpha_k$, математические ожидания $\mu_k$, стандартные отклонения $\sigma_k$, а также точка $x$. Для случайной величины $Y=\sum_{k=1}^{N}\alpha_kX_k$, где $X_k\sim N(\mu_k,\sigma_k^2)$, необходимо с помощью `scipy.stats.norm` вычислить значение функции распределения $F_Y(x)$ и вывести его.

```python
from scipy.stats import norm
```

```python
def main():
    # Считывание данных
    n = int(input())
    mixes = [[float(_) for _ in input().split(" ")
           ] for i in range(n)]
    x = float(input())

    # Функция распределения смеси
    cdf_mix = sum(mix[0] * norm.cdf(x, loc=mix[1], scale=mix[2]) 
                  for mix in mixes)

    print(cdf_mix)


if __name__ == '__main__':
    main()
```

### Задание 1. Маргинальные и условные вероятности

На вход подается таблица совместного распределения дискретных величин $X$ и $Y$ размера $m\times n$ и индексы $i,j$. Необходимо по таблице найти маргинальные вероятности $P_X(x_i)$ и $P_Y(y_j)$ суммированием соответствующей строки и столбца, а также условную вероятность $P(Y=y_j\mid X=x_i)=\frac{P_{X,Y}(x_i,y_j)}{P_X(x_i)}$. Если $P_X(x_i)=0$, вместо условной вероятности вывести `NaN`. Все числовые результаты вывести через пробел с абсолютной погрешностью не более $10^{-6}$.

```python
def main():
    # Считывание данных
    [m, n] = [int(_) for _ in input().split(" ")]
    matrix = [[float(_) for _ in input().split(" ")] for i in range(m)]
    [x, y] = [int(_) for _ in input().split(" ")]

    # Px(x_i) - сумма по y строки с элементом x_i
    Px = sum(matrix[x])
    # Py(y_i) - сумма по x столбца с элементом y_i
    Py = sum(row[y] for row in matrix)
    # P(Y = y_i | X = x_i) - значение элемента (x_i, y_i) / Px(x_i)
    P_margin = matrix[x][y] / Px if Px > 0 else "Nan"

    print(f"{Px} {Py} {P_margin}")


if __name__ == '__main__':
    main()
```

### Задание 2. Математическое ожидание функции двух случайных величин

На вход подаются значения дискретных случайных величин $X$ и $Y$, их совместное распределение $P_{X,Y}$ и коэффициенты $a,b,c,d$. Необходимо вычислить математическое ожидание функции $f(X,Y)=aX+bY+cXY+d$ по всем парам значений, взвешивая каждое значение функции соответствующей совместной вероятностью. Результат вывести с абсолютной погрешностью не более $10^{-6}$.

```python
def main():
    # Считывание данных
    [m, n] = [int(_) for _ in input().split(" ")]
    x = [float(_) for _ in input().split(" ")]
    y = [float(_) for _ in input().split(" ")]
    P = [[float(_) for _ in input().split(" ")] for i in range(m)]
    [a, b, c, d] = [float(_) for _ in input().split(" ")]

    # Математическое ожидание
    result = 0

    for i in range(m):
        for j in range(n):
            f = a * x[i] + b * y[j] + c * x[i] * y[j] + d
            result += f * P[i][j]

    print(result)


if __name__ == '__main__':
    main()
```

### Задание 3. Дисперсии суммы и разности

На вход подаются значения дискретных случайных величин $X$ и $Y$ и их совместное распределение $P_{X,Y}$. Необходимо по таблице вероятностей вычислить дисперсии случайных величин $X+Y$ и $X-Y$. Вывести оба значения через пробел с абсолютной погрешностью не более $10^{-6}$.

```python
def main():
    # Считывание данных
    [m, n] = [int(_) for _ in input().split(" ")]
    x = [float(_) for _ in input().split(" ")]
    y = [float(_) for _ in input().split(" ")]
    P = [[float(_) for _ in input().split(" ")] for i in range(m)]

    # Var(X+Y) = Var(X) + Var(Y) + 2Cov(X,Y)
    # Var(X) = D[X^2] - (D[X])^2
    # Cov(X,Y) = E[XY] - E[X] * E[Y]

    #Вводим все мат. ожидания, которые необходимо рассчитать
    EX = EY = EX2 = EY2 = EXY = 0

    # Рассчитываем мат. ожидания
    for i in range(m):
        for j in range(n):
            EX += x[i] * P[i][j]
            EX2 += pow(x[i], 2) * P[i][j]
            EY += y[j] * P[i][j]
            EY2 += pow(y[j], 2) * P[i][j]
            EXY += x[i] * y[j] * P[i][j]

    # Дисперсия по X
    Var_X = EX2 - pow(EX, 2)
    # Дисперсия по Y
    Var_Y = EY2 - pow(EY, 2)
    # Ковариация
    Cov_XY = EXY - EX * EY

    # Дисперсия суммы
    Var_sum = Var_X + Var_Y + 2*Cov_XY
    # Дисперсия разности
    Var_diff = Var_X + Var_Y - 2*Cov_XY

    print(f"{Var_sum} {Var_diff}")


if __name__ == '__main__':
    main()
```

### Задание 4. Непрерывное совместное распределение

На вход подаются параметры $a,b,c$ плотности $p_{X,Y}(x,y)=K(ax+by+c)$ на единичном квадрате и точки $u,v$. Необходимо определить нормирующую константу $K$, вероятность попадания в область $X\leq u,\;Y\leq v$, а также значения маргинальных плотностей $p_X(u)$ и $p_Y(v)$. Вывести четыре полученных значения через пробел с абсолютной погрешностью не более $10^{-6}$.

```python
def main():
    # Считывание данных
    a, b, c, u, v = map(float, input().split())

    # Нормировка
    # int_0_1 int_0_1 K(ax+by+c)dxdy = 1
    # K (a/2 + b/2 + c) = 1
    # K = 1/ (a/2 + b/2 + c) = 2/(a + b + 2c)
    K = 2 / (a + b + 2 * c)

    # Вероятность попадания в область
    # P_uv =  int_0_u int_0_v K(ax+by+c)dxdy
    # P_uv = K ((au^2v/2) + (buv^2)/2 + cuv)
    P_uv = K * (a*pow(u, 2)*v / 2 + b*u*pow(v, 2) / 2 + c*u* v)

    # Маргинальные плотности
    # Px(u) = int_0_1 K(au+by+c)dy
    # Px(u) = K(au + b/2 + c)
    pX_u = K * (a * u + b / 2 + c)
    # Py(v) = int_0_1 K(ax+bv+c)dx
    # Py(v) = K(a/2 + bv + c)
    pY_v = K * (a / 2 + b * v + c)
    
    print(K, P_uv, pX_u, pY_v)


if __name__ == '__main__':
    main()
```

### Задание 5. Выборочная корреляция Пирсона

На вход подается выборка из $n$ пар $(x_i,y_i)$. Необходимо вычислить коэффициент корреляции Пирсона $r$ между двумя наборами данных. Если все значения хотя бы одного из признаков совпадают, корреляция не определена и следует вывести `NaN`. В остальных случаях вывести $r$ с абсолютной погрешностью не более $10^{-6}$.

```python
def main():
    # Считывание данных
    n = int(input())
    x, y = zip(*(map(float, input().split()) for _ in range(n)))

    # Среднее значение по x и y
    x_mean = sum(x) / n
    y_mean = sum(y) / n

    # Изначально даем коэффициенту Пирсона значение NaN
    r = "NaN"
    # Если дисперсия не равна 0 (все xi одинаковы или все yi одинаковы), считаем
    if not all(xi == x[0] for xi in x) and not all(yi == y[0] for yi in y):
        # Числитель: сумма (xi - x_mean)(yi - y_mean)
        num = sum(map(lambda xi,yi: (xi - x_mean)*(yi - y_mean), x, y))
        # Знаменатель: корень из суммы (xi - x_mean)^2 * корень из суммы (yi - y_mean)^2
        denum_x = math.sqrt(sum(pow(xi - x_mean, 2) for xi in x))
        denum_y = math.sqrt(sum(pow(yi - y_mean, 2) for yi in y))
        # Расчет коэффициента Пирсона
        r = num / (denum_x * denum_y)

    print(r)


if __name__ == '__main__':
    main()
```

### Задание 6. Минимальная дисперсия линейной комбинации

На вход подаются дисперсии случайных величин $X$ и $Y$ и их ковариация: $v_X$, $v_Y$, $c_{XY}$. Для комбинации $Z=wX+(1-w)Y$ необходимо найти вес $w_\ast$, минимизирующий $\operatorname{Var}(Z)$ без ограничений на $w$, и соответствующую дисперсию. Также определить оптимальный вес $w\in[0,1]$ путем ограничения найденного значения диапазоном $[0,1]$ и вычислить минимальную дисперсию для него. Учесть вырожденные случаи, когда знаменатель формулы для $w_\ast$ равен нулю или одна из дисперсий равна нулю. Вывести через пробел $w_\ast$, $\operatorname{Var}(Z)$, ограниченный оптимальный вес и соответствующую дисперсию с абсолютной погрешностью не более $10^{-6}$.

```python
def main():
    # Считывание данных
    vX, vY, cXY = map(float, input().split())

    # Знаменатель критической точки
    check = vX + vY - 2*cXY

    # Если знаменатель = 0 и vX = vY = cXY
    if check == 0 and vX == vY == cXY:
        # Var(Z) не зависит от w
        w_star = 0
    # Если одна из дисперсий нулевая
    elif vX == 0 or vY == 0:
        # то оптимум при w = 1
        w_star = 1
    # иначе: рассчитываем w
    else:
        w_star = (vY - cXY) / check

    # Минимум при ограничении w ∈ [0,1]:
    w_star_clamped = min(max(w_star, 0), 1)
    # Минимальная дисперсия при w*
    var_star = (pow(w_star, 2)*vX + pow(1 - w_star, 2)*vY 
                + 2*w_star*(1-w_star)*cXY)
    # Минимальная дисперсия при w*[0,1]
    var_star_clamped = (pow(w_star_clamped, 2)*vX 
                        + pow(1 - w_star_clamped, 2)*vY 
                        + 2*w_star_clamped*(1-w_star_clamped)*cXY)

    print(f"{w_star} {var_star} {w_star_clamped} {var_star_clamped}")


if __name__ == '__main__':
    main()
```

### Задание 1. Энтропия и перплексия распределения

На вход подается дискретное распределение вероятностей $p_1,\ldots,p_n$, допускающее нулевые значения. Необходимо вычислить энтропию $H(X)=-\sum_i p_i\log_2 p_i$ и перплексию $\operatorname{Perp}(X)=2^{H(X)}$. Для нулевых вероятностей считать вклад в энтропию равным нулю. Вывести $H$ и `Perp` через пробел с абсолютной погрешностью не более $10^{-6}$.

```python
def main():
    # Считывание данных
    n = int(input())
    p = map(float, input().split())

    # Энтропия
    # Допущение: 0log0 = 0
    H = -sum(pi*math.log2(pi) if pi > 0 else 0 
             for pi in p)
    # Перплексия
    Perp = pow(2, H)

    print(f"{H} {Perp}")


if __name__ == '__main__':
    main()
```

### Задание 2. Кросс-энтропия и KL-дивергенция

На вход подаются два дискретных распределения $P$ и $Q$ над одним набором исходов. Необходимо вычислить кросс-энтропию $H(P,Q)$ и KL-дивергенцию $D_{KL}(P\|Q)$. Если для некоторого $i$ выполняется $p_i>0$ и $q_i=0$, обе величины считаются бесконечными и следует вывести `INF`. При $p_i=0$ соответствующий вклад в суммы считать равным нулю. В остальных случаях вывести $H_{PQ}$ и $KL_{PQ}$ через пробел с абсолютной погрешностью не более $10^{-6}$.

```python
def main():
    # Считывание данных
    n = int(input())
    P = [float(_) for _ in input().split(" ")]
    Q = [float(_) for _ in input().split(" ")]

    # Если pi > 0 и qi = 0, то результат бесконечен (INF)
    if any(P[i] > 0 and Q[i] == 0 for i in range(n)):
        print("INF")
    else:
        # Энтропия
        H_PQ = -sum(map(lambda p,q: 
                        p*math.log2(q) if p > 0 else 0, 
                        P, Q))
        # Перплексия
        KL_PQ = sum(map(lambda p,q: 
                        p*math.log2(p/q) if p > 0 else 0, 
                        P, Q))

        print(f"{H_PQ} {KL_PQ}")


if __name__ == '__main__':
    main()
```
