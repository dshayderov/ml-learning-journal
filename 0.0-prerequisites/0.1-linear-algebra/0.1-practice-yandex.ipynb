```python
def cosine_similarity(vec1: np.array, vec2: np.array) -> float:
    """Вычисляет косинусную меру двух векторов."""
    return np.dot(vec1, vec2) / (np.linalg.norm(vec1) * np.linalg.norm(vec2))

def evaulate_result(embeddings: np.array, result_embedding: np.array) -> None:
    """Вычисляет косинусное сходство результата с каждым словом в словаре."""
    print("Косинусное сходство результата с каждым словом:")
    for word, vec in embeddings.items():
		    print(f"{word}: {cosine_similarity(result_embedding, vec):.3f}")

# Пример 1: ebook ≈ book - pages + digital
embeddings = {
    "book":     np.array([0.5, 0.7, 0.6, 0.4, 0.3]),
    "pages":    np.array([0.2, 0.3, 0.1, 0.0, 0.2]),
    "digital":  np.array([0.1, 0.2, 0.4, 0.3, 0.4]),
    "ebook":    np.array([0.3, 0.6, 1.0, 0.7, 0.5]),
    "magazine": np.array([0.3, 0.5, 0.5, 0.2, 0.1])
}

result = embeddings["book"] - embeddings["pages"] + embeddings["digital"]
evaulate_result(embeddings, result)
```

### Задание 1. Линейная комбинация векторов

На вход программы подаются **k** вещественных коэффициентов и **k** векторов одинаковой размерности. Требуется вычислить их линейную комбинацию по формуле $$v=\sum_{i=1}^{k}\lambda_i v_i$$ и вывести координаты полученного вектора. Результат необходимо представить с точностью до **двух знаков после запятой**.

```python
def main():
    k = int(input())
    alphas = [float(alpha) for alpha in input().split(" ")]
    if len(alphas) != k:
        raise Exception("Ввведено неверное количество чисел")
    vectors = [
        [float(alpha) for alpha in input().split(" ")] for n in range(k)
    ]
    if all(len(vec_len) == len(vectors[0]) for vec_len in vectors) == False:
       raise Exception("Ввведены векторы разной длины")

    vec_len = len(vectors[0])
    result = [
        round(
            sum(alphas[i] * vectors[i][j] for i in range(k)), 2
        )
        for j in range(vec_len)
    ]
    
    print(*result)

    
if __name__ == '__main__':
    main()
```

### Задание 2. Проверка ортогональности векторов

На вход программы подаются два целочисленных вектора одинаковой размерности. Необходимо определить, являются ли они ортогональными. Если векторы ортогональны, вывести `ORTHOGONAL`, иначе — `NON-ORTHOGONAL`.

```python
def main():
    n = int(input())
    u = [int(x_u) for x_u in input().split(" ")]
    v = [int(x_v) for x_v in input().split(" ")]

    if not ((0 < n <= 100) or all(x_u <= 1000 for x_u in u) or all(x_v <= 1000 for x_v in u)):
        raise Exception("Ограничения задания не соблюдены")
    if not (len(u) == n and len(v) == n):
        raise Exception("Длина векторов не соответствует заявленной")

    scalar_product = sum(u[i] * v[i] for i in range(n))
    u_norm = pow(sum(pow(x_u, 2) for x_u in u) , 0.5)
    v_norm = pow(sum(pow(x_v, 2) for x_v in v) , 0.5)

    cos_uv = scalar_product / (u_norm * v_norm)

    ort = ""
    if cos_uv == 0:
        ort = "ORTHOGONAL"
    else:
        ort = "NON-ORTHOGONAL"

    print(ort)

    
if __name__ == '__main__':
    main()
```

### Задание 3. Линейная комбинация и базис

На вход программы подаются три целочисленных двумерных вектора. Необходимо определить, можно ли представить третий вектор в виде линейной комбинации первых двух с целочисленными коэффициентами. Если такое представление существует, вывести найденные коэффициенты, иначе — вывести `NO_SOLUTION`.

```python
def main():
    n = 2
    v1 = [int(x_v1) for x_v1 in input().split(" ")]
    v2 = [int(x_v2) for x_v2 in input().split(" ")]
    v3 = [int(x_v3) for x_v3 in input().split(" ")]

    if not (len(v1) == n and len(v2) == n and len(v3) == n):
        raise Exception("Длина векторов не соответствует заявленной")

    alphas = []
    for i in range(10, -10, -1):
        found = False
        for j in range(10, -10, -1):
            if all(i * v1[k] + j * v2[k] == v3[k] for k in range(n)):
                alphas = alphas + [i, j]
                found = True
                break
        if found: break

    if len(alphas) > 0:
        print(*alphas)
    else:
        print("NO_SOLUTION")

    
if __name__ == '__main__':
    main()
```

### Задание 4. Угол между векторами

На вход программы подаются два целочисленных вектора одинаковой размерности. Требуется вычислить угол между ними в градусах и вывести результат в виде целого числа, отбросив дробную часть.

```python
def main():
    n = int(input())
    u = [int(x_u) for x_u in input().split(" ")]
    v = [int(x_v) for x_v in input().split(" ")]

    if not ((0 < n <= 100) or all(x_u <= 100 for x_u in u) or all(x_v <= 100 for x_v in u)):
        raise Exception("Ограничения задания не соблюдены")
    if not (len(u) == n and len(v) == n):
        raise Exception("Длина векторов не соответствует заявленной")

    scalar_product = sum(u[i] * v[i] for i in range(n))
    u_norm = pow(sum(pow(x_u, 2) for x_u in u) , 0.5)
    v_norm = pow(sum(pow(x_v, 2) for x_v in v) , 0.5)

    cos_uv = scalar_product / (u_norm * v_norm)
    angle = int(math.degrees(math.acos(cos_uv)))

    print(angle)

    
if __name__ == '__main__':
    main()
```

### Задание 5. Проверка линейной зависимости векторов

На вход программы подаются **m** целочисленных векторов размерности **n**. Необходимо определить, являются ли заданные векторы линейно зависимыми. Если зависимость существует, вывести `LINEARLY_DEPENDENT`, иначе — `LINEARLY_INDEPENDENT`.

```python
def main():
    [m, n] =[int(par) for par in input().split(" ")]
    if not (2 <= m <= 10 and 2 <= n <= 10):
        raise Exception("Введены неверные параметры m и n")
        
    vectors = [[int(x) for x in input().split(" ")] for vec in range(m)]

    if all(len(vec_len) == n for vec_len in vectors) == False:
       raise Exception("Длина векторов не соответствует заявленной")

    matrix = np.array(vectors)
    r = np.linalg.matrix_rank(matrix)
    status = None

    if r < m:
        status = "LINEARLY_DEPENDENT"
    elif r == m:
        status = "LINEARLY_INDEPENDENT"

    print(status)

    
if __name__ == '__main__':
    main()
```
