# Линейная алгебра для нейросетей. Векторы.

## Подготовка:

```python
import numpy as np
from numpy.linalg import norm
```

```python
# Списки сущностей:
names = {
    0: "Элерон'Таз",
    1: "Чилит-в-Глубинах",
    2: "Ротгарн Стопа",
    3: "Ульфгард Золотая Струя",
    4: "Мурка",
    5: "Красная машина",
    6: "Носит-под-Гузник",
    7: "Куремар",
}

vec_list = np.array([
    [3, 9, 3, 0, 8, 2, 0, 4, 2],        # Элерон'Таз
    [2, 7, 4, 9, 8, 3, 0, 5, 4],        # Чилит-в-Глубинах
    [9, 3, 2, 0, 1, 8, 0, 5, 9],        # Ротгарн Стопа
    [7, 2, 3, 7, 4, 6, 0, 2, 6],        # Ульфгард Золотая Струя
    [5, 8, 1, 7, 2, 9, 0, 2, 3],        # Мурка
    [25, 10, 10, 0, 20, 30, 0, 0, 15],  # Красная машина
    [3, 3, 3, 3, 3, 3, 3, 3, 3],        # Носит-под-Гузник
    [1, 3, 9, 5, 1, 1, 0, 9, 1],        # Куремар
])

mods = [
    2.0,  # Зелье божественной силы
    1.2,  # Зелье могущества
    0.5,  # Сонливость
]

potions = np.array([
    [3, 0, 0, 0, 0, 0, 0, 0, 0],      # Зелье силы
    [2, 1, 0, 0, 0, 1, 0, 1, 1],      # Зелье война
    [-2, -1, 0, 0, 0, 0, 4, 0, -1],   # Яд паралича
    [0, 0, 3, 0, 0, 0, 0, 0, -1],     # Зелье ученого
    [1, 1, -1, 0, 0, 3, 0, 0, 1],     # Скама
    [0, 0, -2, -1, -1, 0, 0, 0, 1],   # Яд безумия
    [-2, 0, 0, 3, 1, 0, 0, -2, 0],    # Зелье красноречия
])

diseases = np.array([
    [-2, -2, -2, 0, -1, -3, 0, 1, 0],  # Простуда
    [1, 0, -1, -1, -1, 3, 9, 0, 0],    # Каменная хворь
    [-3, -1, 0, 1, -5, 0, 0, 3, 0],    # Чума
])
```

```python
# Таблица героев:
print(f'{'Имя:':<25} {'Очки навыков:':<15} {'Мощь:':<10}')
for i in range(len(vec_list)):
    points = norm(vec_list[i], ord=1)
    power = round(norm(vec_list[i], ord=2), 2)
    print(f'{names[i]:<25} {points:<15} {power:<10}')
```

## Примеры операций:

### Умножение на скаляр:

```python
result = vec_list[0] * mods[1]
print(f"{vec_list[0]} * 1.3 = {result}")
```

### Сложение векторов:

```python
result = vec_list[0] + potions[0]
print(f"{vec_list[0]} + {potions[0]} = {result}")
```

### Норма:

```python
result = round(np.linalg.norm(vec_list[0], ord=2), 2)
print(f"{vec_list[0]} + {potions[0]} = {result}")
```

### Скалярное произведение векторов:

```python
p_norm1_0 = norm(vec_list[0], ord=2)
p_norm1_1 = norm(vec_list[1], ord=2)
result = vec_list[0] @ vec_list[1] / (p_norm1_0 * p_norm1_1)
result = round(result, 2)

print(f"{vec_list[0]} * {vec_list[1]} = {result}")
```

### Векторное произведение векторов:

```python
a = vec_list[0][:3]
b = vec_list[1][:3]
result = np.cross(a, b)

print(f"{a} * {b} = {result}")
```

### Задание 1. Магический эксперимент.

Куремар решил провести эксперимент по расширению своих возможностей. Для этого он употребил все зелья из доступных (potions) и, как побочный эффект, получил модификатор сонливости.<p>
Вычислите изменение евклидовой нормы героя, чтобы узнать, стал ли он сильнее.

```python
# Вводные:
kuremar = vec_list[7].copy()
sleepiness = mods[2]
print(kuremar)
```

```python
# Решение:
start_norm = round(np.linalg.norm(kuremar, ord=2), 2)               # норма Куремара до принятия зелий
experiment = (kuremar + sum(potions)) * sleepiness
new_norm = round(np.linalg.norm(experiment, ord=2), 2)              # норма Куремара после принятия зелий 
effect = new_norm - start_norm                                      # полученный эффект

print(f"Евклидова норма Куремара до эксперимента: {start_norm}")
print(f"Евклидова норма Куремара после эксперимента: {new_norm}")
print(f"Изменение евклидовой нормы героя: {effect}")
```

### Задание 2. Вызов на бой.

**Можно применить лишь по одному модификатору из каждого списка на одного героя.**<p>

```python
# Вводные:
nappy = vec_list[6].copy()
red_machine = vec_list[5].copy()
print(nappy, red_machine)
```

```python
worst_potion = min([sum(potion) for potion in potions])                        # зелье с худшим эффектом 
best_potion = max([sum(potion) for potion in potions])                         # зелье с лучшим эффектом

worst_disease = min([sum(disease) for disease in diseases])                    # болезнь с худшим эффектом
best_disease = max([sum(disease) for disease in diseases])                     # болезнь с лучшим эффектом

worst_mod = min(mods)                                                          # состояние с худшим эффектом
best_mod = max(mods)                                                           # состояние с лучшим эффектом

nappy_effect = round(np.linalg.norm((nappy + best_potion + 
                                     best_disease) * best_mod, ord=2), 2)      # лучшие модификаторы - Носит-под-Гузнику
red_effect = round(np.linalg.norm((red_machine + worst_potion + 
                                   worst_disease) * worst_mod, ord=2), 2)      # худшие - Красной машине

print(f"Евклидова норма Носит-под-Гузника: {nappy_effect}")
print(f"Евклидова норма Красной машины: {red_effect}")

if nappy_effect > red_effect: 
    print("Носит-по-Гузник одерживает победу")
else: 
    print("Красную машину не победить")
```

Настало время финальной битвы! Герои прибыли на знаменитую арену и разбились на команды. Все, кроме Красной машины, которого оштрафовали за слишком высокий параметр дерзости (его вектор = -1).
Перед тем как начнется битва, участникам предстоит сразиться с великаном, мощь которого равна 60. Те, кто одолеет испытание, пройдут в финальный раунд.

Разъяснение. Используя векторное умножение и евклидову норму, найдите все возможные пары героев и их суммарную мощь. Отсейте тех, кто не набрал 60 очков. Выясните, количество уникальных пар, прошедших в финал, а также, какая команда станет победителем соревнования.

Подведите итоги, сделав информативную статистику:

1. **Основные итоги.** Выведите топ уникальных пар, прошедших в финал в порядке убывания;

2. **Аутсайдеры.** Отдельным списком выведите тех, кто не прошел в финал ни в одной из команд;

3. **Линый результат.** Для каждого героя найдите сумму очков всех команд в которых он принимал участие. Выведите личный вклад героев (включая аутсайдеров) в порядке убывания.

```python
# Вводные:
vec_short_list = vec_list[:, :3].copy()
vec_short_list[5] = [-1, -1, -1]
threshold = norm([60, 0, 0], ord=2)

print(f"Мощь великана: {threshold}\n")
print(f"Таблица героев")
print(f'{'Имя:':<25}{'Сила:':<15}{'Ловкость:':<15}{'Интеллект:':<15}')
for i, row in enumerate(vec_short_list):
    print(f'{names[i]:<25}{row[0]:<15}{row[1]:<15}{row[2]:<15}')
```

```python
# Решение:
n = len(vec_short_list)                                                               # количество персонажей
pair_list = []                                                                        # список всех возможных пар

for i in range(n-1):
    for j in range(i+1, n):
        char_a = vec_short_list[i]
        char_b = vec_short_list[j]
        pair_power = round(norm(np.cross(char_a, char_b), ord=2), 2)
        pair_list.append([names[i], names[j], pair_power])                            # наполнение списка

print(f"Все возможные пары и их мощь:")
print(f'{'Персонаж 1:':<25}{'Персонаж 2:':<25}{'Суммарная мощь:':<15}')
for row in pair_list:
    print(f'{row[0]:<25}{row[1]:<25}{row[2]:<15}')

finalist_pair_count = len([pair for pair in pair_list if pair[2] > threshold])        # количество пар, прошедших порог мощи
print(f"\nКоличество уникальных пар, прошедших в финал: {finalist_pair_count}")

winners = max(pair_list, key=lambda row: row[2])                                      # наимощнейшая пара
print(f"\nПобедители соревнования: {winners[0]} и {winners[1]}. Их мощь: {winners[2]}")

print(f"\n\nИтоги:\n")

top_finalists = sorted([pair for pair in pair_list if pair[2] > threshold],           # пары, прошедшие в финал
                       key=lambda row: row[2], reverse=True)                          # сортировка по мощи в порядке убывания
print("1. Топ уникальных пары, прошедших в финал:")
print(f'{'Персонаж 1:':<25}{'Персонаж 2:':<25}{'Суммарная мощь:':<15}')
for row in top_finalists:
    print(f'{row[0]:<25}{row[1]:<25}{row[2]:<15}')

outsiders = [
    [name,                                                                            # имя аутсайдера
     max(row[2] for row in pair_list if name in (row[0],row[1]))                      # максимальная мощь пары, в которой был аутсайдер
         ] for name in names.values() 
      if any(name in (row[0], row[1]) for row in top_finalists) == False              # отбор по персонажам, которых нет в списке финалистов
]
print("\n2. Аутсайдеры:")
print(f'{'Персонаж:':<25}{'Максимальная мощь пары:':<15}')
for row in outsiders:
    print(f'{row[0]:<25}{row[1]:<15}')

individual = sorted([
    [name,                                                                            # имя персонажа
     round(sum(row[2] for row in pair_list if name in (row[0],row[1])), 2)            # суммарная мощь пар, в которых был персонаж
         ] for name in names.values()
], key=lambda row: row[1], reverse=True)                                              # сортировка по мощи в порядке убывания
print("\n3. Личный вклад героев:")
print(f'{'Персонаж:':<25}{'Личный вклад:':<15}')
for row in individual:
    print(f'{row[0]:<25}{row[1]:<15}')
```
