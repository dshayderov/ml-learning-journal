### Задача 2

Вычислите скалярное произведение векторов `a = [1, 2, 3]` и `b = [4, -5, 6]` вручную (без `np.dot`), используя цикл или список, затем проверьте результат с помощью `np.dot`.

````python
a = [1, 2, 3]
b = [4, -5, 6]

scalar_product = sum(a[i] * b[i] for i in range(len(a)))
check = np.dot(a, b)

print(f"Скалярное произведение вручную: {scalar_product}")
print(f"Проверка через np.dot: {check}")
````

### Задача 3

Найдите евклидову норму (длину) вектора `v = [3, 4, 12]`.

````python
v = [3, 4, 12]
norm = math.sqrt(sum(pow(x, 2) for x in v))
check = np.linalg.norm(v, ord=2)

print(f"Норма вектора вручную: {norm}")
print(f"Проверка через numpy: {check}")
````

### Задача 4

Даны векторы `a = [1, 0, 0]`, `b = [0, 1, 0]`. Вычислите угол между ними в градусах по формуле `cos θ = (a·b)/(|a||b|)`.

````python
a = [1, 0, 0]
b = [0, 1, 0]

scalar_product = np.dot(a, b)
a_norm = np.linalg.norm(a, ord=2)
b_norm = np.linalg.norm(b, ord=2)

cos_θ = scalar_product / (a_norm * b_norm)
angle_θ = math.degrees(math.acos(cos_θ))

print(f"Угол между векторами: {angle_θ}")
````

### Задача 5

Нормализуйте вектор `v = [6, 8]` (приведите к единичной длине). Проверьте, что длина результата равна 1.

````python
v = [6, 8]
norm = np.linalg.norm(v, ord=2)

v_normalize = np.array([x/norm for x in v])
check = np.linalg.norm(v_normalize, ord=2)

print(f"Нормализованный вектор: {v_normalize}")
print(f"Проверка длины: {check}")
````

### Задача 6

Даны векторы `a = [2, 3]` и `b = [4, -1]`. Найдите проекцию вектора `a` на вектор `b` (формула: `proj = (a·b) / |b|`).

````python
a = [2, 3]
b = [4, -1]

scalar_product = np.dot(a, b)
b_norm = np.linalg.norm(b, ord=2)

proj = round(scalar_product / b_norm, 3)

print(f"Проекция ветора a на вектор b: {proj}")
````

### Задача 20

Найдите собственные значения и собственные векторы матрицы `A = [[2,0],[0,3]]` с помощью `np.linalg.eig`. В комментарии объясните, почему для диагональной матрицы собственные векторы совпадают с осями координат.

````python
A = np.array([
    [2, 0],
    [0, 3]
])

eigenvalues, eigenvectors = np.linalg.eig(A)

print("Собственные значения: ", eigenvalues)
print("Собственные векторы:\n", eigenvectors)
````

Для диагональной матрицы A = diag(2, 3):

A @ [1, 0] = 2 * [1, 0]

A @ [0, 1] = 3 * [0, 1]

Поэтому векторы [1, 0] и [0, 1] являются собственными.
Они совпадают с направлениями осей координат.

### Задача 25

Реализуйте функцию `matrix_power(A, n)`, которая возводит квадратную матрицу в натуральную степень `n` через последовательное умножение (без `np.linalg.matrix_power`). Проверьте на `A=[[1,1],[0,1]]`, `n=5`.

````python
def matrix_power(A: np.array, n: int) -> np.array:
    result = A.copy()
    for i in range(n - 1):
        result = result @ A

    return np.array(result)

A = np.array([[1,1],[0,1]])
n = 5

A_pow = matrix_power(A, n)
A_pow_check = np.linalg.matrix_power(A, n)

print("Возведение матрицы в степень через функцию\n", A_pow)
print("Возведение матрицы в степень через Numpy\n", A_pow_check)
````
