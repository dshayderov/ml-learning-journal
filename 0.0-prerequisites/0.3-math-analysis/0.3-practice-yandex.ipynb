### Задание 2. Проверка непрерывности функции

На вход подаются функция $f(x)$ в виде строки, точка $x_0$ и малое положительное число $\delta$. Необходимо определить непрерывность функции в точке $x_0$, сравнив значения $f(x_0-\delta)$ и $f(x_0+\delta)$ с $f(x_0)$ с допуском $\varepsilon=5\delta$. Если оба отклонения меньше $\varepsilon$, вывести `CONTINUOUS`, иначе — `DISCONTINUOUS`.

>*Версия с библиотекой `sympy`*

```python
import sympy as sp
```

```python
def main():
    # Считывание функции, заданной в виде строки
    x = sp.Symbol('x')
    expr = sp.sympify(input())
    func = sp.lambdify(x, expr)

    # Ввод x0 и delta
    x0 = float(input())
    delta = float(input())

    # Допуск
    k = 5
    epsilon = k * delta

    # Проверка на непрерывность
    cont = False
    if abs(func(x0 - delta) - func(x0)) < epsilon:
        if abs(func(x0 + delta) - func(x0)) < epsilon:
            cont = True

    # Вывод результата
    if cont is True:
        print("CONTINUOUS")
    else:
        print("DISCONTINUOUS")


if __name__ == '__main__':
    main()
```

### Задание 3. Липшицева непрерывность

На вход подаются функция $f(x)$, интервал $[a,b]$ и положительная константа $L$. Необходимо проверить, выполняется ли для всех точек интервала условие Липшица с константой $L$: $|f(x_1)-f(x_2)|\leq L|x_1-x_2|$. Если условие выполняется, вывести `LIPSCHITZ`, иначе — `NOT LIPSCHITZ`.

```python
import math
```

```python
def main():
    e = math.e
    
    # Считывание функции, заданной в виде строки
    f_expr = input()
    f = eval('lambda x, e=e: ' + f_expr)

    # Ввод интервала и константы
    interval = [float(x) for x in input().split()]
    L = eval(input())

    # Проверка на непрерывность
    if (abs(f(interval[0]) - f(interval[1])) < 
        L * abs(interval[0] - interval[1])):
        print("LIPSCHITZ")
    else:
        print("NOT LIPSCHITZ")


if __name__ == '__main__':
    main()
```

### Задание 1. Нахождение критических точек функции

На вход подаются коэффициенты $a,b,c,d$ кубического полинома и границы интервала $[p,q]$. Необходимо найти все точки внутри интервала, в которых $f'(x)=0$, определить характер каждой точки — локальный максимум, локальный минимум или точка перегиба — и вычислить значение функции в них. Критические точки вывести в порядке возрастания $x$, округляя координаты и значения функции до 5 знаков после запятой. Если таких точек нет, вывести `No critical points found.`.

>*Версия с библиотекой `sympy`*

```python
def main():    
    # Считывание данных
    [a, b, c, d] = [float(_) for _ in input().split()]
    [p, q] = [float(_) for _ in input().split()]

    x = sp.symbols('x')
    f = a * x**3 + b * x**2 + c * x + d

    # Поиск критических точек
    # Находим первую производную
    f_prime = sp.diff(f, x)

    # Приравниваем производную к нулю и решаем уравнение
    critical_points = sp.solve(f_prime, x)

    # Фильтруем точки, оставляя только те, что внутри интервала
    points_inside = [point for point in critical_points if p < point < q]

    if len(points_inside) == 0:
        # Критических точек нет
        print("No critical points found.")
    else:
        # Критические точки есть
        for point in points_inside:
            # Вторая производная
            f_prime_2nd = sp.diff(f, x, 2)
            #Значение второй производной в точке
            f_2nd_res = f_prime_2nd.subs(x, point)

            # Характер критической точки
            if f_2nd_res > 0:
                print("Local minimum", end=" ")
            elif f_2nd_res < 0:
                print("Local maximum", end=" ")
            else:
                print("Saddle point", end=" ")

            print(f"at x = {point:.5f}")

            # Значение функции в точке
            func_res = f.subs(x, point)
            print(f"f(x) = {func_res:.5f}")   


if __name__ == '__main__':
    main()
```

>*Версия без библиотеки `sympy`*

```python
def main():    
    # Считывание данных
    [a, b, c, d] = [float(_) for _ in input().split()]
    [p, q] = [float(_) for _ in input().split()]
    
    # Выводим первую производную в виде квадратного уравнения
    # f'(x) = 3ax^2 + 2bx + c
    A = 3*a
    B = 2*b
    C = c

    # Список корней
    roots = []

    if A == 0:
        # Линейное уравнение Bx + C = 0
        if B != 0:
            if (-C) % B == 0:
                roots.append((-C) // B)
    else:
        # Квадратное уравнение Ax^2 + Bx + C = 0
        D = B ** 2 - 4 * A * C

        if D >= 0:
            sqrt_D = math.sqrt(D)

            roots.append((-B + sqrt_D) / (2 * A))
            
            if D != 0:
                roots.append((-B - sqrt_D) / (2 * A))

    # Фильтрация по интервалу
    roots = [root for root in roots if p <= root <= q]
    
    if roots:
        roots.sort()
        # Критические точки есть
        for x in roots:
            # Вторая производная
            # f''(x) = 6ax + 2b
            diff_2nd = 6 * a * x + 2 * b
            # Характер критической точки
            if diff_2nd > 0:
                print("Local minimum", end=" ")
            elif diff_2nd < 0:
                print("Local maximum", end=" ")
            else:
                print("Saddle point", end=" ")

            print(f"at x = {x:.5f}")

            # Значение функции в точке
            f = a * x**3 + b * x**2 + c * x + d
            print(f"f(x) = {f:.5f}")
    else:
        # Критических точек нет
        print("No critical points found.")


if __name__ == '__main__':
    main()
```

### Задание 2. Нахождение корня методом Ньютона

На вход подаются коэффициенты квадратной функции $f(x)=ax^2+bx+c$, начальное приближение $x_0$ и допустимая погрешность $\varepsilon$. Необходимо реализовать метод Ньютона по формуле $x_{n+1}=x_n-\frac{f(x_n)}{f'(x_n)}$ и найти значение $x$, для которого $|f(x)|<\varepsilon$. Если корень найден не более чем за 1000 итераций, вывести его с точностью до 6 знаков после запятой и количество выполненных итераций. Если метод не сходится за заданное число итераций или производная становится равной нулю, вывести `Solution not found`.

```python
def main():   
    e = math.e
    
    # Считывание данных
    [a, b, c] = [float(_) for _ in input().split()]
    x0 = float(input())
    epsilon = eval(input())
    
    x_n = x0
    it = 0
    solted = False

    for i in range(1, 1001):
        # Вычисление значения функции
        f = a * x_n**2 + b * x_n + c
        # Проверка на выполнение условия
        if abs(f) < epsilon:
            solted = True
            # Если значение функции удовлетворяет условию, значит
            # решение было найдено на предыдущей итерации
            it = i - 1
            break
        else:
            # Вычисление производной
            f_diff = 2*a*x_n + b
            if f_diff == 0:
                break
            # Новый шаг по методу Ньютона
            x_n = x_n - f / f_diff
       
    if solted:
        print(f"Root found: x = {x_n:.6f}")
        print(f"Number of iterations: {it}")
    else:
        print("Solution not found")

if __name__ == '__main__':
    main()
```

### Задание 1. Выбор направления шага

На вход подаются градиент $g$ в точке $x_0$, $k$ возможных направлений $v_i$ и длина шага $\varepsilon$. Для каждого направления необходимо учитывать его единичную версию и по линейному приближению оценить изменение функции $\Delta L$ после шага длины $\varepsilon$. Требуется выбрать направление, обеспечивающее наибольшее уменьшение функции, и вывести его индекс (с единицы) и соответствующее $\Delta L$. При равных результатах выбирается направление с наименьшим индексом. Значение $\Delta L$ вывести с точностью до 4 знаков после запятой.

```python
def main():       
    # Считывание первой строки
    first_input = input().split(" ")
    [n, k] = [int(_) for _ in first_input[0:-1]]
    epsilon = float(first_input[2])

    # Градиент в точке x0
    g = [float(_) for _ in input().split()]
    # k векторов направлений
    vectors = [[float(_) for _ in input().split()] for i in range(k)]

    # Массив изменений функции
    deltas = []
    for v in vectors:
        # Нормируем вектор направления (приводим к единичной длине)
        norm = math.sqrt(sum(x**2 for x in v))
        u = [x / norm for x in v]

        # Оцениваем изменение функции
        delta = epsilon * sum(g[i] * u[i] for i in range(n))

        deltas.append(delta)

    # Выбираем направление, которое дает наибольшее уменьшение функции
    best_idx = deltas.index(min(deltas))
    print(f"{best_idx + 1} {deltas[best_idx]:.4f}")
    

if __name__ == '__main__':
    main()
```

### Задание 3. Проверка аналитического градиента

На вход подаются матрица $A$, векторы $b$ и $w$, а также параметры $\lambda$, $\alpha$ и шаг $h$. Необходимо вычислить аналитический градиент функции, включающей квадратичную ошибку $\|Aw-b\|^2$, L2-регуляризацию $\lambda\|w\|^2$ и нелинейную составляющую $\alpha\sum_j\sin(w_j)$, а затем получить численный градиент по каждой координате методом центральной разности с шагом $h$. Требуется найти максимальное абсолютное расхождение между соответствующими координатами аналитического и численного градиентов и вывести его с точностью до 6 знаков после запятой.

```python
import numpy as np
```

```python
def L(w, A, b, lam, alpha):
    """Исследуемая функция"""
    error = A @ w - b

    # Квадратичная ошибка
    square_error = np.sum(error ** 2)
    # L2-регуляризация
    l2_reg = lam * np.sum(w ** 2)
    # Нелинейная составляющая
    unlinear = alpha * np.sum(np.sin(w))

    return (square_error + l2_reg + unlinear)
    

def main():       
    # Считывание данных
    [m, n] = [int(_) for _ in input().split(" ")]
    A = np.array([[float(_) for _ in input().split(" ")] for i in range(m)])
    b = np.array([float(_) for _ in input().split(" ")])
    w = np.array([float(_) for _ in input().split(" ")])
    [lam, alpha, h] = [float(_) for _ in input().split(" ")]

    # Аналитический градиент
    # g' = (||Aw - b||^2)' + (lam||w||^2)' + (alpha*sum(sin(w)))' =
    # = 2A^T(Aw - b) + 2*lam*w + alpha*cos(w)
    g = 2*A.T@(A @ w - b) + 2*lam*w + alpha*np.cos(w)

    # Расчетный градиент
    # g_j = (L(w_j + h) - L(w_j - h)) / 2h
    g_num = []

    for j in range(len(w)):
        w_plus = w.copy()
        w_minus = w.copy()
    
        w_plus[j] += h
        w_minus[j] -= h
    
        g_j = (L(w_plus, A, b, lam, alpha) - L(w_minus, A, b, lam, alpha)) / (2 * h)
    
        g_num.append(g_j)
    
    g_num = np.array(g_num)

    # Вычисляем максимальное расхождение по координатам
    # max |g - g_num|
    res = np.max(np.abs(g - g_num))

    print(f"{res:.6f}")
    

if __name__ == '__main__':
    main()
```
